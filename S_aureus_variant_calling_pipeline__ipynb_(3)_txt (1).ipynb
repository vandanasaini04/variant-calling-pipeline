```python
%%bash

RAW="/content/variants/USA300_ERR17521341.raw.vcf"
FILTERED="/content/variants/USA300_ERR17521341.filtered.vcf"

python3 - "$RAW" "$FILTERED" <<'PY'
import sys

raw = sys.argv[1]
filtered = sys.argv[2]

total = 0
kept = 0

with open(raw) as fin, open(filtered, "w") as fout:
    for line in fin:
        if line.startswith("#"):
            fout.write(line)
            continue

        total += 1
        fields = line.rstrip("\n").split("\t")

        qual = float(fields[5]) if fields[5] != "." else 0

        info = {}
        for item in fields[7].split(";"):
            if "=" in item:
                key, value = item.split("=", 1)
                info[key] = value

        af = float(info["AF"].split(",")[0])
        dp = float(info["DP"])

        # ORIGINAL PIPELINE CRITERIA
        if qual > 1000 and af == 1 and dp > 10:
            fout.write(line)
            kept += 1

print("=== FILTERING COMPLETE ===")
print("Raw variants:", total)
print("High-confidence variants:", kept)
print("Output:", filtered)
PY
```

```python
%%bash

FILTERED="/content/variants/USA300_ERR17521341.filtered.vcf"

echo "=== FILTERED VCF VALIDATION ==="

echo "Total filtered variants:"
grep -v '^#' "$FILTERED" | wc -l

echo
echo "Checking that all variants satisfy:"
echo "QUAL > 1000"
echo "AF = 1"
echo "DP > 10"

python3 - "$FILTERED" <<'PY'
import sys

vcf = sys.argv[1]
total = 0
bad = 0

for line in open(vcf):
    if line.startswith("#"):
        continue

    total += 1
    fields = line.rstrip().split("\t")

    qual = float(fields[5])

    info = {}
    for item in fields[7].split(";"):
        if "=" in item:
            k, v = item.split("=", 1)
            info[k] = v

    af = float(info["AF"])
    dp = int(float(info["DP"]))

    if not (qual > 1000 and af == 1 and dp > 10):
        bad += 1

print(f"Total variants checked: {total}")
print(f"Variants failing criteria: {bad}")

if bad == 0:
    print("PASS: all filtered variants satisfy the criteria.")
else:
    print("WARNING: some variants do not satisfy the criteria.")
PY
```

```python
# Annotate 57 variants against GFF3 using Python
import re

gff_file = "/content/reference/NC_007793.1.gff3"
vcf_file = "/content/variants/USA300_ERR17521341.filtered.vcf"
out_file = "/content/variants/USA300_ERR17521341.annotated.tsv"

# Parse gene features from GFF3
genes = []
with open(gff_file) as f:
    for line in f:
        if line.startswith("#"):
            continue
        fields = line.strip().split("\t")
        if len(fields) < 9:
            continue
        if fields[2] == "gene":
            start = int(fields[3])
            end = int(fields[4])
            attrs = fields[8]
            name_match = re.search(r'Name=([^;]+)', attrs)
            id_match = re.search(r'ID=([^;]+)', attrs)
            gene_name = name_match.group(1) if name_match else "unknown"
            gene_id = id_match.group(1) if id_match else "unknown"
            genes.append((start, end, gene_name, gene_id))

def find_gene(pos):
    pos = int(pos)
    for start, end, name, gid in genes:
        if start <= pos <= end:
            return name, gid
    return "intergenic", "NA"

# Annotate variants
results = []
with open(vcf_file) as f:
    for line in f:
        if line.startswith("#"):
            continue
        fields = line.strip().split("\t")
        chrom = fields[0]
        pos = fields[1]
        ref = fields[3]
        alt = fields[4]
        qual = fields[5]
        info = fields[7]

        dp = af = "NA"
        for item in info.split(";"):
            if item.startswith("DP="):
                dp = item.split("=")[1]
            elif item.startswith("AF="):
                af = item.split("=")[1]

        gene_name, gene_id = find_gene(pos)
        results.append([chrom, pos, ref, alt, qual, dp, af, gene_name, gene_id])

# Save table
with open(out_file, "w") as f:
    f.write("CHROM\tPOS\tREF\tALT\tQUAL\tDP\tAF\tGENE\tGENE_ID\n")
    for r in results:
        f.write("\t".join(r) + "\n")

# Summary
import pandas as pd
df = pd.read_csv(out_file, sep="\t")
print(f"Total annotated variants: {len(df)}")
print(f"\nIntergenic variants: {(df['GENE'] == 'intergenic').sum()}")
print(f"Genic variants: {(df['GENE'] != 'intergenic').sum()}")
print("\nGenes with variants:")
print(df[df['GENE'] != 'intergenic'][['POS','REF','ALT','QUAL','GENE']].to_string(index=False))
display(df)
```
